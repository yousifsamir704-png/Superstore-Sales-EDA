## Question 1

What is the total sales generated by the company?

## Question 2
What is the total profit generated by the company?

```python
print(f'Total Profits${df['Profit'].sum()}')
```

### Insight
The company generated a total profit of approximately $286,397. This indicates that while total sales exceeded $2.29 million, only a portion of revenue was converted into profit.

## Question 3

### Insight

### Insight

## Question 5
What is the average sales per order?

```python
Total_sales = df['Sales'].sum()
Total_Order = df['Order ID'].nunique()
Sales_per_order = (Total_sales/Total_Order)
print(f"Avg Sales per order :${Sales_per_order}" )
```

### Insight
On average, each order generated approximately $458.61 in sales.


## Question 6
Which category generates the highest sales?

```python
df.groupby(['Category'])['Sales'].sum().sort_values(ascending=False).plot(kind="bar")
```

### Insight
Technology generated the highest sales among all categories, making it the company's strongest revenue driver.


## Question 7
Which category generates the highest profit?


```python
df.groupby(['Category'])['Profit'].sum().plot(kind='barh')
```

### Insight
Technology generated the highest profit among all categories, while Furniture contributed the lowest profit. This suggests that Technology products are the most profitable segment for the business.


## Question 8
Which sub-category generates the highest sales?

```python
df.groupby(['Sub-Category'])['Sales'].sum().sort_values(ascending=False).plot(kind='barh')
```

### Insight
Phones generated the highest sales among all sub-categories, followed by Chairs and Storage. This indicates that these products are the primary drivers of revenue.

## Question 9
Which sub-category generates the highest profit?

```python
df.groupby(['Sub-Category'])['Profit'].sum().sort_values(ascending=False).plot(kind='barh')
```

### Insight
Copiers generated the highest Profit among all sub-categories, followed by Phones and Accessorise. This indicates that these products are the primary drivers of revenue.

## Question 10
Which region generates the highest sales?

```python
df.groupby(['Region'])['Sales'].sum().sort_values(ascending=False).plot(kind='pie')
```

### Insight
The West region generated the highest sales among all regions, followed by the East region. The South region contributed the lowest sales.

## Question 11
Does a higher discount lead to lower profit?

```python
df[['Discount','Profit']].corr()
```

```python
df.plot(kind='scatter',x='Discount',y='Profit')
```

### Insight
The correlation between Discount and Profit is -0.219, indicating a weak negative relationship. This suggests that higher discounts tend to reduce profitability, although the relationship is not strong enough to conclude that discounts are the sole factor affecting profit.


## Question 11
Which category has the highest profit margin?

```python
category_margin = df.groupby('Category')[['Sales','Profit']].sum()
category_margin['Porfit_margin'] = (
category_margin['Profit'] /
category_margin['Sales']
) * 100
round(category_margin['Porfit_margin'],2).plot(kind='bar')
```

## Insight
Technology achieved the highest profit margin, making it the most efficient category in converting sales into profit.
